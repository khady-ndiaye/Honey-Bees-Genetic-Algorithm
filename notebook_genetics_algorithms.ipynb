##  VEILLE TECHNOLOGIQUE: Algorithmes Génétiques (AG)

Les **algorithmes génétiques (AG)** sont des méthodes d’optimisation inspirées de la théorie de l’évolution de Darwin (sélection naturelle, croisement, mutation).
Ils sont particulièrement efficaces pour résoudre des problèmes complexes où les méthodes classiques échouent ou coûtent trop cher (comme le **problème du voyageur de commerce (TSP)**, l’ordonnancement de tâches, la conception de réseaux de neurones, etc.).


##  Étapes principales d’un Algorithme Génétique

### 1. **Initialisation de la population**

L’AG commence avec une population initiale d’individus (solutions candidates).

#### Méthodes :

* **Aléatoire (Random Initialization)**
  * Chaque individu est généré de manière aléatoire.
  *   Facile, génère une grande diversité.
  *   Peut contenir beaucoup de solutions médiocres.
* **Heuristique (Heuristic Initialization)**
  * Utilisation d’algorithmes gloutons (ex: nearest-neighbor pour TSP).
  *   Population initiale plus compétitive.
  *   Moins de diversité → risque de stagnation.
* **Hybride (Mixed Initialization)**
  * Mélange aléatoire + heuristique.
  * Bon compromis entre diversité et qualité initiale.



### 2. **Évaluation (Fitness Function)**

Chaque individu reçoit un score (fitness) représentant sa performance.

#### Méthodes :

* **Fitness directe**
  * La fonction objectif est utilisée telle quelle (ex: minimiser distance totale).
* **Fitness inversée**
  * Exemple : fitness = `1 / distance`.
  * Évite des valeurs de fitness trop grandes.
* **Fitness normalisée**
  * La fitness est transformée pour que la somme des scores = 1.
  * Favorise une sélection équilibrée.
* **Fitness multi-objectifs** (MOEA, NSGA-II)
  * Prend en compte plusieurs critères (ex: minimiser distance ET maximiser diversité).



### 3. **Sélection**

Les parents sont choisis parmi la population pour se reproduire.

#### Méthodes :

* **Roulette Wheel (Proportionnelle)**
  * Probabilité de sélection proportionnelle à la fitness.
  *   Simple, favorise les bons individus.
  *   Si une solution domine, perte de diversité.
* **Tournoi (Tournament Selection)**
  * On choisit k individus au hasard → le meilleur gagne.
  *   Contrôle via la taille du tournoi.
  *   Risque de convergence prématurée si k trop grand.
* **Rang (Rank Selection)**
  * Les individus sont triés par fitness, puis la probabilité dépend du rang.
  *   Évite la domination d’un seul individu.
  *   Plus lent que la roulette.
* **Élitisme (Elitism)**
  * Conservation automatique des meilleurs individus.
  *   Empêche la perte des meilleures solutions.
  *   Peut réduire la diversité si utilisé seul.



### 4. **Croisement (Crossover)**

Combine deux parents pour créer de nouveaux enfants.

#### Méthodes générales :

* **One-Point Crossover**
  * Coupure en un point → segments échangés.
  *   Rapide, simple.
  *   Risque de perdre la structure des gènes.
* **Two-Point Crossover**
  * Deux points de coupure → segments centraux échangés.
  *   Plus de diversité que le one-point.
  *   Toujours risque de casser des structures utiles.
* **Uniform Crossover**
  * Chaque gène choisi au hasard entre les deux parents.
  *  Très diversifiant.
  *  Peut produire des enfants incohérents (doublons).

#### Méthodes spécialisées pour les permutations (comme TSP) :

* **Order Crossover (OX)**
  * Un segment du parent1 est copié, puis le reste est rempli avec l’ordre du parent2.
  *  Préserve l’ordre relatif.
  *  Dépend fortement de la taille du segment choisi.
* **Partially Mapped Crossover (PMX)**
  * Échange d’un segment + mapping des gènes.
  *  Préserve positions et valeurs → très populaire pour TSP.
  *  Un peu plus complexe à coder.
* **Cycle Crossover (CX)**
  * Les enfants sont construits en suivant des cycles entre parents.
  *  Chaque gène garde sa position relative.
  *  Complexe et parfois peu diversifiant.
* **Edge Recombination (ERX)**
  * Utilise les arêtes (relations entre gènes) des parents pour construire l’enfant.
  *  Très adapté aux problèmes de graphe.
  *  Implémentation plus lourde.



### 5. **Mutation**

Introduit de la diversité en modifiant un individu.

#### Méthodes pour permutations :

* **Swap Mutation** : échange deux positions.
* **Insertion Mutation** : déplace un gène ailleurs.
* **Inversion Mutation** : inverse une sous-séquence.
* **Scramble Mutation** : mélange aléatoire d’une sous-séquence.

#### Méthodes pour bitstring :

* **Flip Mutation** : change un bit (0 ↔ 1).

   Mutation faible → exploration progressive.  
   Mutation trop forte → perte de convergence.



### 6. **Remplacement**

Création de la nouvelle génération à partir de la population et des enfants.

#### Méthodes :

* **Generational Replacement**
  * Toute la population est remplacée par les enfants.
  *   Très exploratoire.
  *   Risque de perdre les meilleurs.
* **Élitisme**
  * Les meilleurs individus sont gardés → complétés par les enfants.
  *   Sécurité → ne jamais perdre les bonnes solutions.
  *   Peut réduire la diversité.
* **Steady-State**
  * Remplacement partiel (quelques individus).
  *   Plus stable, convergence progressive.
  *   Moins diversifiant.



### 7. **Critère d’arrêt**

Détermine quand l’AG doit s’arrêter.

#### Méthodes :

* **Nombre fixe de générations**.
* **Convergence** : si la fitness n’évolue plus depuis X générations.
* **Temps limite**.
* **Seuil de fitness** : si une solution atteint une valeur cible.



##  Schéma global d’un Algorithme Génétique

```mermaid
flowchart TD
    A[Initialisation de la population] --> B[Évaluation Fitness]
    B --> C[Sélection]
    C --> D[Croisement]
    D --> E[Mutation]
    E --> F[Évaluation des enfants]
    F --> G[Remplacement]
    G --> H{Critère d'arrêt ?}
    H -- Oui --> I[Meilleure solution]
    H -- Non --> C
```



##  Points clés pour bien implémenter un AG

* **Population** : assez grande pour la diversité, pas trop pour le coût.
* **Taux de mutation** : typiquement entre `0.01` et `0.1`.
* **Taux de crossover** : généralement haut (`0.6` à `0.9`).
* **Élitisme** recommandé pour conserver les meilleures solutions.
* **Choix du crossover** :
  * Bitstring → One-point, Uniform.
  * Permutations (TSP) → OX, PMX, ERX.
* **Diversité** : essentielle pour éviter la convergence prématurée.



##  Pseudo-code Python d’un Algorithme Génétique

```python
def genetic_algorithm():
    # 1. Initialisation
    population = init_population(size)
    
    for generation in range(max_generations):
        # 2. Évaluation
        fitness_scores = [evaluate(ind) for ind in population]
        
        # 3. Sélection
        parents = selection(population, fitness_scores)
        
        # 4. Croisement
        offspring = []
        for i in range(0, len(parents), 2):
            child1, child2 = crossover(parents[i], parents[i+1])
            offspring.append(child1)
            offspring.append(child2)
        
        # 5. Mutation
        offspring = [mutate(child, mutation_rate) for child in offspring]
        
        # 6. Remplacement
        population = replacement(population, offspring, fitness_scores)
        
        # 7. Critère d’arrêt (optionnel : convergence, seuil, etc.)
        if stop_condition_met(population, fitness_scores):
            break
    
    best_solution = max(population, key=lambda ind: evaluate(ind))
    return best_solution
```

````python
# Imports
import random
import math
import pandas as pd
import matplotlib.pyplot as plt
import networkx as nx
from dataclasses import dataclass, field
from typing import List, Tuple, Optional

plt.rcParams['figure.figsize'] = (8,6)
````

````python
@dataclass
class Bee:
    id: int
    path: List[Tuple[float, float]]
    parents: Optional[Tuple[int, int]] = None
    fitness: Optional[float] = None
    distance: Optional[float] = None

    def evaluate(self, hive_position: Tuple[float, float]):
        dist = 0.0
        current = hive_position
        for flower in self.path:
            dist += math.dist(current, flower)
            current = flower
        dist += math.dist(current, hive_position)
        self.distance = dist
        # évite division par zéro
        self.fitness = 1.0 / dist if dist > 0 else float('inf')
        return self.fitness

class Beehive:
    def __init__(self, flowers: List[Tuple[float,float]], hive_position=(500,500),
                 population_size=100, mutation_rate=0.05, generations=100):
        self.flowers = flowers
        self.hive_position = hive_position
        self.population_size = population_size
        self.mutation_rate = mutation_rate
        self.generations = generations
        self.population: List[Bee] = []
        self.genealogy = {}  # id -> (parents, generation)
        self.history: List[float] = []    # temps moyen par génération
        self._id_counter = 0

    def _next_id(self):
        i = self._id_counter
        self._id_counter += 1
        return i

    def init_population(self):
        self.population = []
        for _ in range(self.population_size):
            path = self.flowers[:]
            random.shuffle(path)
            bee = Bee(self._next_id(), path, parents=None)
            bee.evaluate(self.hive_position)
            self.population.append(bee)
            self.genealogy[bee.id] = (bee.parents, 0)

    def crossover_ox(self, parent1: Bee, parent2: Bee) -> Bee:
        # OX order crossover
        n = len(parent1.path)
        a, b = sorted(random.sample(range(n), 2))
        child_path = [None]*n
        child_path[a:b] = parent1.path[a:b]
        p2_genes = [g for g in parent2.path if g not in child_path]
        idx = 0
        for i in range(n):
            if child_path[i] is None:
                child_path[i] = p2_genes[idx]; idx += 1
        return Bee(self._next_id(), child_path, parents=(parent1.id, parent2.id))

    def mutate(self, bee: Bee):
        if random.random() < self.mutation_rate:
            i, j = random.sample(range(len(bee.path)), 2)
            bee.path[i], bee.path[j] = bee.path[j], bee.path[i]

    def evolve(self):
        self.init_population()
        for gen in range(1, self.generations+1):
            # évaluation
            for bee in self.population:
                bee.evaluate(self.hive_position)

            avg_time = sum(b.distance for b in self.population)/len(self.population)
            self.history.append(avg_time)

            # tri et sélection élitiste : garder les top (population_size//2)
            self.population.sort(key=lambda b: b.fitness, reverse=True)
            survivors = self.population[: self.population_size // 2]

            # reproduction uniquement parmi les survivors (50 meilleurs)
            children = []
            while len(survivors) + len(children) < self.population_size:
                p1, p2 = random.sample(survivors, 2)
                child = self.crossover_ox(p1, p2)
                self.mutate(child)
                child.evaluate(self.hive_position)
                self.genealogy[child.id] = (child.parents, gen)
                children.append(child)

            self.population = survivors + children

        best = max(self.population, key=lambda b: b.fitness)
        return best
````

````python
def load_flowers_from_csv(csv_path: str):
    df = pd.read_csv(csv_path)
    # Accepte différentes colonnes courantes
    if 'x' in df.columns and 'y' in df.columns:
        return list(zip(df['x'], df['y']))
    #  coordonnées 'X','Y' 
    for a,b in [('X','Y')]:
        if a in df.columns and b in df.columns:
            return list(zip(df[a], df[b]))
    # sinon assume deux premières colonnes sont coordonnées
    return list(zip(df.iloc[:,0], df.iloc[:,1]))



````

````python
def plot_best_path(best_bee: Bee, hive_position=(500,500)):
    x = [hive_position[0]] + [p[0] for p in best_bee.path] + [hive_position[0]]
    y = [hive_position[1]] + [p[1] for p in best_bee.path] + [hive_position[1]]
    plt.figure()
    plt.plot(x,y, marker='o')
    plt.scatter([hive_position[0]],[hive_position[1]], marker='s', s=100, label='Ruche')
    plt.title('Trajet de la meilleure abeille')
    plt.legend()
    plt.show()

def plot_history(beehive: Beehive):
    plt.figure()
    plt.plot(beehive.history)
    plt.xlabel('Génération'); plt.ylabel('Temps moyen'); plt.title('Évolution du temps moyen')
    plt.show()

def plot_genealogy(beehive: Beehive, best_bee: Bee, max_depth=6):
    G = nx.DiGraph()
    stack = [(best_bee.id,0)]
    visited = set()
    while stack:
        bee_id, depth = stack.pop()
        if bee_id in visited or depth>max_depth:
            continue
        visited.add(bee_id)
        parents, gen = beehive.genealogy.get(bee_id, (None,None))
        if parents:
            for p in parents:
                if p is not None:
                    G.add_edge(p, bee_id)
                    stack.append((p, depth+1))
    plt.figure(figsize=(10,6))
    pos = nx.spring_layout(G)
    nx.draw(G, pos, with_labels=True, node_size=300, font_size=8, arrows=True)
    plt.title("Arbre généalogique (meilleure abeille)")
    plt.show()
````

````python

if __name__ == '__main__':
    
    flowers = load_flowers_from_csv("C:/Users/ndiay/OneDrive/Desktop/miel_abeille/venv/Champ_fleurs.csv")
    hive_pos = (500,500)
    hive = Beehive(flowers, hive_pos, population_size=40, mutation_rate=0.05, generations=80)
    best = hive.evolve()
    print(f"Meilleure abeille id={best.id}, distance={best.distance:.2f}")
    plot_best_path(best, hive_pos)
    plot_history(hive)
    plot_genealogy(hive, best)
````

````python
def compare_parameters(flowers, hive_pos):
    configs = [0.01, 0.05, 0.1]
    for mr in configs:
        hive = Beehive(flowers, hive_pos, population_size=40, mutation_rate=mr, generations=80)
        hive.evolve()
        plt.plot(hive.history, label=f'mutation={mr}')
    plt.xlabel('Générations'); plt.ylabel('Temps moyen'); plt.legend(); plt.title('Comparaison taux de mutation')
    plt.show()


````
